# CA on the children dataset

Classic correspondence analysis on a contingency table.

```python
import matplotlib.pyplot as plt
from factominer import CA
from factominer.datasets import load_children

ch = load_children()
print(ch.head())
```

```python
res = CA(ch, ncp=4, row_sup=list(range(14, 18)), col_sup=list(range(5, 8)))
print(res.eig.round(4))
```

```python
from factominer.plot import plot
fig, ax = plt.subplots(figsize=(7, 6))
plot(res, choix="biplot", ax=ax)
plt.show()
```
